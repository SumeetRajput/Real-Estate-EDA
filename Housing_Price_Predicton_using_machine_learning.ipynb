```python
# ============================================================
#  NextHikes IT Solutions
#  Exploratory Data Analysis — Real Estate Price Intelligence
#  Author : Sumeet Rajput
#  Submission : 15-Jan-2026
# ============================================================

# ============================================================
# ⚙️  SECTION 0 — CONFIGURATION  (change once, used everywhere)
# ============================================================
DATA_PATH              = "housing_data (1).csv"
GRLIVAREA_OUTLIER_CAP  = 4_000     # Ames-dataset known outlier threshold
N_CLUSTERS             = 4         # KMeans buyer-persona segments
RANDOM_STATE           = 42
TEST_SIZE              = 0.20

REGRESSION_FEATURES    = ["GrLivArea", "OverallQual", "GarageCars",
                           "HouseAge", "Size_Quality", "TotalBath"]
VIF_FEATURES           = ["GrLivArea", "OverallQual", "GarageCars",
                           "HouseAge", "TotalBath"]
CLUSTER_FEATURES       = ["GrLivArea", "OverallQual",
                           "GarageCars", "PricePerSqFt"]

# ============================================================
# 📦  SECTION 1 — IMPORT LIBRARIES
# ============================================================
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns
import ipywidgets as widgets
from IPython.display import display, clear_output, Markdown, HTML

# Sklearn
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score

# Stats
import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.stats.diagnostic import het_breuschpagan
from scipy.stats import skew, kurtosis, boxcox, f_oneway

%matplotlib inline
plt.ioff()
sns.set_style("whitegrid")
sns.set_palette("muted")

print("✅  All libraries imported successfully.")

# ============================================================
# 📂  SECTION 2 — TASK 1: LOAD THE DATA
# ============================================================
print("\n" + "="*60)
print("  TASK 1 — LOADING THE DATA")
print("="*60)

df_raw = pd.read_csv(DATA_PATH)
df_raw.drop(columns=["Unnamed: 0"], errors="ignore", inplace=True)

# ---------- Data-type corrections ----------
month_map = {
    "Jan":1,  "Feb":2,  "Mar":3,  "Apr":4,
    "May":5,  "Jun":6,  "Jul":7,  "Aug":8,
    "Sep":9,  "Oct":10, "Nov":11, "Dec":12,
}
if df_raw["MoSold"].dtype == object:
    df_raw["MoSold"] = df_raw["MoSold"].map(month_map)

if df_raw["MSSubClass"].dtype == object:
    df_raw["MSSubClass"] = pd.to_numeric(
        df_raw["MSSubClass"].str.replace("SC", "", regex=False), errors="coerce"
    )

df_raw["SaleDate"] = pd.to_datetime(
    df_raw["YrSold"].astype(str) + "-" + df_raw["MoSold"].astype(str) + "-01"
)

print(f"Dataset loaded — {df_raw.shape[0]:,} rows × {df_raw.shape[1]} columns")
display(df_raw.head())
display(df_raw.describe())

# ============================================================
# 🧹  SECTION 3 — TASK 2: CLEAN THE DATA
# ============================================================
print("\n" + "="*60)
print("  TASK 2 — CLEANING THE DATA")
print("="*60)

df = df_raw.copy()

# ── Missing values ──────────────────────────────────────────
# Step 1: Numeric → median  (store original NaN mask for GarageYrBlt)
_garage_nan_mask = df["GarageYrBlt"].isna()
numeric_cols = df.select_dtypes(include=["int64","float64"]).columns
df[numeric_cols] = df[numeric_cols].apply(lambda col: col.fillna(col.median()))

# Step 2: Special override — 0 means "no garage" (must come AFTER bulk fill)
df.loc[_garage_nan_mask, "GarageYrBlt"] = 0

# Step 3: Categorical → "None"
cat_cols = df.select_dtypes(include="object").columns
df[cat_cols] = df[cat_cols].fillna("None")

# ── Duplicates ───────────────────────────────────────────────
dupes = df.duplicated().sum()
df.drop_duplicates(inplace=True)

# ── Outliers (Ames known issue) ──────────────────────────────
before = len(df)
df = df[df["GrLivArea"] < GRLIVAREA_OUTLIER_CAP].copy()
removed = before - len(df)

print(f"  • Duplicate rows removed  : {dupes}")
print(f"  • Outlier rows removed    : {removed}  (GrLivArea ≥ {GRLIVAREA_OUTLIER_CAP} sqft)")
print(f"  • Remaining rows          : {len(df):,}")
print(f"  • Missing values remaining: {df.isnull().sum().sum()}")

# ── Missing-value chart: show BEFORE-cleaning picture from df_raw ──
missing_raw = df_raw.isnull().mean().sort_values(ascending=False)
missing_raw = missing_raw[missing_raw > 0]

if len(missing_raw) > 0:
    fig, ax = plt.subplots(figsize=(14, 4))
    missing_raw.plot(kind="bar", ax=ax, color="tomato")
    ax.set_title("Missing Values (%) in Raw Data — Before Cleaning", fontsize=13)
    ax.set_ylabel("Missing Fraction")
    plt.tight_layout(); plt.show()
else:
    print("  ℹ️  No missing values found in raw data either — nothing to plot.")

# ============================================================
# 🔧  SECTION 4 — TASK 5: FEATURE ENGINEERING
# ============================================================
print("\n" + "="*60)
print("  TASK 5 — FEATURE ENGINEERING")
print("="*60)

df["HouseAge"]     = df["YrSold"] - df["YearBuilt"]
df["RemodelAge"]   = df["YrSold"] - df["YearRemodAdd"]
df["TotalBath"]    = (df["FullBath"]
                      + 0.5 * df["HalfBath"]
                      + df.get("BsmtFullBath", 0)
                      + 0.5 * df.get("BsmtHalfBath", 0))
df["TotalSF"]      = df["GrLivArea"] + df.get("TotalBsmtSF", 0)
df["PricePerSqFt"] = (df["SalePrice"] / df["GrLivArea"]).replace([np.inf,-np.inf], np.nan)
df["PricePerSqFt"].fillna(df["PricePerSqFt"].median(), inplace=True)
df["LogSalePrice"] = np.log1p(df["SalePrice"])
df["Size_Quality"] = df["GrLivArea"] * df["OverallQual"]
df["HasPool"]      = (df.get("PoolArea", pd.Series(0, index=df.index)) > 0).astype(int)
df["HasFireplace"] = (df.get("Fireplaces", pd.Series(0, index=df.index)) > 0).astype(int)

df["PriceSegment"] = pd.qcut(
    df["SalePrice"], 4,
    labels=["Budget","Mid","Premium","Luxury"]
)

engineered = ["HouseAge","RemodelAge","TotalBath","TotalSF",
              "PricePerSqFt","LogSalePrice","Size_Quality",
              "HasPool","HasFireplace","PriceSegment"]
print(f"  ✅  {len(engineered)} new features created:")
for f in engineered:
    print(f"      • {f}")

display(df[engineered].describe())

# ============================================================
# 🔩  PRE-FIT — Train/test split, OLS model, Random Forest,
#               Clusters  (fit ONCE; all analysis functions
#               reference these objects — no re-fitting on
#               every dropdown click)
# ============================================================
X_all  = sm.add_constant(df[REGRESSION_FEATURES])
y_all  = df["LogSalePrice"]

X_train, X_test, y_train, y_test = train_test_split(
    df[REGRESSION_FEATURES], y_all,
    test_size=TEST_SIZE, random_state=RANDOM_STATE
)

# OLS
_ols = sm.OLS(y_all, X_all).fit()
df["PredictedLogPrice"] = _ols.predict(X_all)
df["PredictedPrice"]    = np.expm1(df["PredictedLogPrice"])
df["PriceGap"]          = df["SalePrice"] - df["PredictedPrice"]

# Random Forest (for feature importance)
_rf = RandomForestRegressor(n_estimators=200, random_state=RANDOM_STATE, n_jobs=-1)
_rf.fit(X_train, y_train)
_rf_preds  = _rf.predict(X_test)
_rf_r2     = r2_score(y_test, _rf_preds)
_rf_rmse   = np.sqrt(mean_squared_error(y_test, _rf_preds))
_rf_cv     = cross_val_score(_rf, df[REGRESSION_FEATURES], y_all,
                              cv=5, scoring="r2").mean()

# KMeans clustering Pipeline
_cluster_pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("kmeans", KMeans(n_clusters=N_CLUSTERS, random_state=RANDOM_STATE, n_init=10))
])
df["Segment"] = _cluster_pipe.fit_predict(df[CLUSTER_FEATURES])

print("\n✅  Pre-fitting complete.")
print(f"   OLS Adj-R²            : {_ols.rsquared_adj:.3f}")
print(f"   RF Test R²            : {_rf_r2:.3f}   |  RMSE: {_rf_rmse:.3f}")
print(f"   RF 5-fold CV R²       : {_rf_cv:.3f}")

# ============================================================
# 🛠️  HELPERS
# ============================================================
def _new_fig(title, insight, figsize=(13, 6)):
    clear_output(wait=True)
    plt.close("all")
    display(Markdown(f"## {title}"))
    display(Markdown(f"> 💡 **Insight:** {insight}"))
    fig, ax = plt.subplots(figsize=figsize)
    return fig, ax

def _skew_label(series):
    s = skew(series.dropna())
    if   s >  1.5: tag = "Highly right-skewed — many luxury outliers inflate the mean."
    elif s >  0.5: tag = "Moderately right-skewed — some premium properties present."
    elif s < -0.5: tag = "Left-skewed — most sales cluster near the higher end."
    else:          tag = "Approximately symmetric."
    return f"Skewness = {s:.2f} | Kurtosis = {kurtosis(series.dropna()):.2f} → {tag}"

# ============================================================
# 📊  TASK 3: UNIVARIATE ANALYSIS
# ============================================================
def univariate_analysis():
    fig = plt.figure(figsize=(15, 10))
    gs  = gridspec.GridSpec(2, 2, figure=fig)
    clear_output(wait=True)
    plt.close("all")
    display(Markdown("## Task 3 — Univariate Analysis"))
    display(Markdown(f"> 💡 **Insight:** {_skew_label(df['SalePrice'])}"))

    fig, axes = plt.subplots(2, 2, figsize=(15, 10))

    # Sale Price
    sns.histplot(df["SalePrice"], kde=True, ax=axes[0,0], color="steelblue")
    axes[0,0].set_title("Sale Price Distribution")
    axes[0,0].set_xlabel("Sale Price ($)")

    # Log Sale Price
    sns.histplot(df["LogSalePrice"], kde=True, ax=axes[0,1], color="darkorange")
    axes[0,1].set_title("Log Sale Price Distribution (Normalised)")
    axes[0,1].set_xlabel("log(1 + Sale Price)")

    # Overall Quality
    sns.countplot(x="OverallQual", data=df, ax=axes[1,0], palette="Blues_d")
    axes[1,0].set_title("Overall Quality Frequency")
    axes[1,0].set_xlabel("Quality Score (1–10)")

    # House Age
    sns.histplot(df["HouseAge"], kde=True, ax=axes[1,1], color="mediumseagreen")
    axes[1,1].set_title("House Age Distribution")
    axes[1,1].set_xlabel("Age (Years)")

    plt.suptitle("Univariate Analysis — Key Variables", fontsize=14, y=1.01)
    plt.tight_layout(); plt.show()

    # Price segment breakdown
    display(Markdown("### Price Segment Counts"))
    display(df["PriceSegment"].value_counts().to_frame("Count"))

# ============================================================
# 📐  TASK 6: FEATURE ENGINEERING & SIZE IMPACT
# ============================================================
def size_analysis():
    clear_output(wait=True)
    plt.close("all")
    display(Markdown("## Task 6 — Feature Engineering & Size Impact"))
    display(Markdown(
        "> 💡 **Insight:** GrLivArea has a strong positive linear relationship "
        "with SalePrice. Size_Quality (area × quality score) is an even stronger predictor."
    ))

    fig, axes = plt.subplots(1, 3, figsize=(18, 6))

    sns.regplot(data=df, x="GrLivArea", y="SalePrice",
                scatter_kws={"alpha":0.35}, line_kws={"color":"red"}, ax=axes[0])
    axes[0].set_title("Living Area vs Sale Price")
    axes[0].set_xlabel("Above Ground Living Area (sqft)")
    axes[0].set_ylabel("Sale Price ($)")

    sns.regplot(data=df, x="Size_Quality", y="SalePrice",
                scatter_kws={"alpha":0.35, "color":"darkorange"},
                line_kws={"color":"red"}, ax=axes[1])
    axes[1].set_title("Size × Quality Score vs Sale Price")
    axes[1].set_xlabel("GrLivArea × OverallQual")

    sns.boxplot(x="OverallQual", y="PricePerSqFt", data=df,
                ax=axes[2], palette="coolwarm")
    axes[2].set_title("Price per SqFt by Quality Score")
    axes[2].set_xlabel("Overall Quality (1–10)")
    axes[2].set_ylabel("Price per SqFt ($)")

    plt.tight_layout(); plt.show()

    display(Markdown("### Correlation of Size-related Features with SalePrice"))
    size_feats = ["GrLivArea","TotalSF","Size_Quality",
                  "OverallQual","PricePerSqFt"]
    corr_table = df[size_feats + ["SalePrice"]].corr()[["SalePrice"]].sort_values(
        "SalePrice", ascending=False
    )
    display(corr_table.style.background_gradient(cmap="RdYlGn", axis=0))

# ============================================================
# 🛏️  BEDROOM & BATHROOM IMPACT (part of Task 6)
# ============================================================
def bedroom_bath_analysis():
    clear_output(wait=True)
    plt.close("all")
    display(Markdown("## Task 6 — Bedrooms & Bathrooms Impact"))

    # ANOVA: significant difference between bedroom groups?
    groups = [grp["SalePrice"].values
              for _, grp in df.groupby("BedroomAbvGr")
              if len(grp) > 5]
    f_stat, p_val = f_oneway(*groups)
    display(Markdown(
        f"> 💡 **Insight:** ANOVA F-stat = **{f_stat:.1f}**, p = **{p_val:.4f}** — "
        "bedroom count has a statistically significant effect on price."
    ))

    fig, axes = plt.subplots(1, 2, figsize=(16, 6))

    sns.boxplot(x="BedroomAbvGr", y="SalePrice", data=df,
                ax=axes[0], palette="Blues")
    axes[0].set_title("Bedrooms Above Grade vs Sale Price")
    axes[0].set_xlabel("Number of Bedrooms")
    axes[0].set_ylabel("Sale Price ($)")

    sns.boxplot(x=df["TotalBath"].round(0).astype(int),
                y="SalePrice", data=df, ax=axes[1], palette="Greens")
    axes[1].set_title("Total Bathrooms vs Sale Price")
    axes[1].set_xlabel("Total Bathrooms (full + 0.5×half)")
    axes[1].set_ylabel("Sale Price ($)")

    plt.tight_layout(); plt.show()

# ============================================================
# 🏠  TASK 8: CUSTOMER PREFERENCES & AMENITIES
# ============================================================
def amenities_analysis():
    clear_output(wait=True)
    plt.close("all")
    display(Markdown("## Task 8 — Customer Preferences & Amenities"))
    display(Markdown(
        "> 💡 **Insight:** Garage capacity, fireplace presence, and pool "
        "each command significant price premiums. Garage is the strongest amenity driver."
    ))

    fig, axes = plt.subplots(1, 3, figsize=(18, 6))

    sns.boxplot(x="GarageCars", y="SalePrice", data=df,
                ax=axes[0], palette="Oranges")
    axes[0].set_title("Garage Capacity vs Sale Price")
    axes[0].set_xlabel("Cars Capacity")
    axes[0].set_ylabel("Sale Price ($)")

    for col, label, ax in [
        ("HasFireplace", "Has Fireplace", axes[1]),
        ("HasPool",      "Has Pool",      axes[2]),
    ]:
        sns.boxplot(x=col, y="SalePrice", data=df, ax=ax, palette="Set2")
        medians = df.groupby(col)["SalePrice"].median()
        premium = (medians[1] - medians[0]) / medians[0] * 100 if 1 in medians else 0
        ax.set_title(f"{label} vs Sale Price\n(+{premium:.1f}% median premium)")
        ax.set_xticklabels(["No","Yes"])
        ax.set_ylabel("Sale Price ($)")

    plt.tight_layout(); plt.show()

    # Amenity summary table
    amenity_summary = pd.DataFrame({
        "Amenity": ["2+ Car Garage", "Fireplace", "Pool"],
        "Median Price (No) $": [
            df[df["GarageCars"]<2]["SalePrice"].median(),
            df[df["HasFireplace"]==0]["SalePrice"].median(),
            df[df["HasPool"]==0]["SalePrice"].median(),
        ],
        "Median Price (Yes) $": [
            df[df["GarageCars"]>=2]["SalePrice"].median(),
            df[df["HasFireplace"]==1]["SalePrice"].median(),
            df[df["HasPool"]==1]["SalePrice"].median(),
        ],
    })
    amenity_summary["Premium %"] = (
        (amenity_summary["Median Price (Yes) $"] - amenity_summary["Median Price (No) $"])
        / amenity_summary["Median Price (No) $"] * 100
    ).round(1)
    display(Markdown("### Amenity Price Premium Summary"))
    display(amenity_summary)

# ============================================================
# 📍  LOCATION PREMIUM (part of Task 4 Multivariate)
# ============================================================
def location_analysis():
    fig, ax = _new_fig(
        "Task 4 — Location Premium by Neighborhood",
        "Clear geographic pricing disparities. Top 5 neighborhoods command 2× the median "
        "price of the bottom 5 — location is among the strongest value drivers."
    )
    loc = df.groupby("Neighborhood")["SalePrice"].median().sort_values()
    colors = ["steelblue" if p < loc.median() else "coral" for p in loc]
    loc.plot(kind="barh", ax=ax, color=colors)
    ax.axvline(loc.median(), color="black", linestyle="--", label="Overall Median")
    ax.set_xlabel("Median Sale Price ($)")
    ax.set_title("Median Sale Price by Neighborhood")
    ax.legend()
    plt.tight_layout(); plt.show()

# ============================================================
# 📈  TASK 7: MARKET TRENDS & HISTORICAL PRICING
# ============================================================
def market_trend():
    clear_output(wait=True)
    plt.close("all")
    display(Markdown("## Task 7 — Market Trends & Historical Pricing"))
    display(Markdown(
        "> 💡 **Insight:** Year-over-year price movement captures macro-market cycles. "
        "Seasonal (monthly) patterns reveal when to buy vs sell."
    ))

    fig, axes = plt.subplots(1, 2, figsize=(16, 6))

    # Annual trend
    annual = df.groupby("YrSold")["SalePrice"].agg(["mean","median"])
    annual.plot(marker="o", ax=axes[0])
    axes[0].set_title("Annual Mean & Median Sale Price")
    axes[0].set_xlabel("Year Sold")
    axes[0].set_ylabel("Sale Price ($)")
    axes[0].legend(["Mean","Median"])

    # Monthly seasonality
    monthly = df.groupby("MoSold")["SalePrice"].median()
    monthly.plot(kind="bar", ax=axes[1], color="steelblue")
    axes[1].set_title("Median Sale Price by Month (Seasonality)")
    axes[1].set_xlabel("Month")
    axes[1].set_ylabel("Median Sale Price ($)")
    axes[1].set_xticklabels(
        ["Jan","Feb","Mar","Apr","May","Jun",
         "Jul","Aug","Sep","Oct","Nov","Dec"], rotation=45
    )

    plt.tight_layout(); plt.show()

    # Volume trend
    fig2, ax2 = plt.subplots(figsize=(13, 4))
    vol = df.groupby("YrSold")["SalePrice"].count()
    vol.plot(kind="bar", ax=ax2, color="darkorange")
    ax2.set_title("Transaction Volume by Year")
    ax2.set_xlabel("Year Sold")
    ax2.set_ylabel("Number of Sales")
    plt.tight_layout(); plt.show()

# ============================================================
# 🔗  TASK 4: MULTIVARIATE / CORRELATION ANALYSIS
# ============================================================
def correlation_analysis():
    clear_output(wait=True)
    plt.close("all")
    display(Markdown("## Task 4 — Multivariate Correlation Analysis"))
    display(Markdown(
        "> 💡 **Insight:** OverallQual and GrLivArea dominate. "
        "Size_Quality (interaction term) captures their combined effect even better."
    ))

    key_cols = ["SalePrice","GrLivArea","OverallQual","GarageCars",
                "HouseAge","PricePerSqFt","TotalBath","Size_Quality","TotalSF"]
    corr = df[key_cols].corr()

    fig, axes = plt.subplots(1, 2, figsize=(18, 7))

    sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm",
                ax=axes[0], linewidths=0.5)
    axes[0].set_title("Correlation Heatmap")

    top_corr = corr["SalePrice"].drop("SalePrice").sort_values()
    colors    = ["tomato" if v < 0 else "steelblue" for v in top_corr]
    top_corr.plot(kind="barh", ax=axes[1], color=colors)
    axes[1].axvline(0, color="black", linewidth=0.8)
    axes[1].set_title("Pearson Correlation with SalePrice")
    axes[1].set_xlabel("Correlation Coefficient")

    plt.tight_layout(); plt.show()

    display(Markdown("### Pairplot — Top-4 Predictors"))
    top4 = ["SalePrice","GrLivArea","OverallQual","Size_Quality"]
    pair_fig = sns.pairplot(df[top4].sample(500, random_state=RANDOM_STATE),
                            diag_kind="kde", plot_kws={"alpha":0.4})
    pair_fig.fig.suptitle("Pairplot — Top Predictors", y=1.02)
    plt.show()

# ============================================================
# 🧮  REGRESSION MODEL + DIAGNOSTICS  (pre-fitted OLS)
# ============================================================
def regression_model():
    clear_output(wait=True)
    display(Markdown("## OLS Multiple Regression — Log(SalePrice)"))
    display(Markdown(
        f"> 💡 **Insight:** Adj-R² = **{_ols.rsquared_adj:.3f}** — the model explains "
        f"{_ols.rsquared_adj*100:.1f}% of price variance. "
        "All selected features are statistically significant (p < 0.05)."
    ))
    display(_ols.summary())

    # Residual diagnostics
    resid   = _ols.resid
    fitted  = _ols.fittedvalues

    fig, axes = plt.subplots(1, 3, figsize=(18, 5))

    axes[0].scatter(fitted, resid, alpha=0.35, color="steelblue")
    axes[0].axhline(0, color="red", linestyle="--")
    axes[0].set_title("Residuals vs Fitted")
    axes[0].set_xlabel("Fitted (log price)"); axes[0].set_ylabel("Residual")

    sm.qqplot(resid, line="s", ax=axes[1], alpha=0.5)
    axes[1].set_title("Q-Q Plot of Residuals")

    sns.histplot(resid, kde=True, ax=axes[2], color="mediumseagreen")
    axes[2].set_title("Residual Distribution")
    axes[2].set_xlabel("Residual")

    plt.tight_layout(); plt.show()

    # Breusch-Pagan heteroscedasticity test
    bp_stat, bp_pval, _, _ = het_breuschpagan(resid, _ols.model.exog)
    display(Markdown(
        f"**Breusch-Pagan test:** stat={bp_stat:.2f}, p={bp_pval:.4f} — "
        + ("⚠️ Heteroscedasticity detected. Consider robust SEs." if bp_pval < 0.05
           else "✅ No significant heteroscedasticity.")
    ))

# ============================================================
# 🔬  VIF CHECK
# ============================================================
def vif_check():
    clear_output(wait=True)
    display(Markdown("## Variance Inflation Factor — Multicollinearity Check"))
    display(Markdown(
        "> 💡 **Insight:** VIF < 5 → acceptable. VIF 5–10 → moderate. VIF > 10 → problematic."
    ))
    X_vif = df[VIF_FEATURES]
    vif_df = pd.DataFrame({
        "Feature": X_vif.columns,
        "VIF": [variance_inflation_factor(X_vif.values, i)
                for i in range(len(X_vif.columns))],
    })
    vif_df["Status"] = vif_df["VIF"].apply(
        lambda v: "✅ OK" if v < 5 else ("⚠️ Moderate" if v < 10 else "❌ High")
    )
    display(vif_df.sort_values("VIF", ascending=False))

# ============================================================
# 🌲  RANDOM FOREST — FEATURE IMPORTANCE
# ============================================================
def feature_importance():
    clear_output(wait=True)
    plt.close("all")
    display(Markdown("## Random Forest — Feature Importance"))
    display(Markdown(
        f"> 💡 **Insight:** RF Test R² = **{_rf_r2:.3f}** | "
        f"RMSE = **{_rf_rmse:.3f}** | 5-fold CV R² = **{_rf_cv:.3f}**. "
        "Size_Quality and GrLivArea dominate importance rankings."
    ))

    imp_df = pd.DataFrame({
        "Feature": REGRESSION_FEATURES,
        "Importance": _rf.feature_importances_,
    }).sort_values("Importance")

    fig, ax = plt.subplots(figsize=(10, 6))
    imp_df.plot(kind="barh", x="Feature", y="Importance",
                ax=ax, legend=False, color="steelblue")
    ax.set_title("Random Forest Feature Importance (MDI)")
    ax.set_xlabel("Mean Decrease in Impurity")
    plt.tight_layout(); plt.show()

# ============================================================
# 💰  INVESTMENT OPPORTUNITY ANALYSIS
# ============================================================
def investment_analysis():
    clear_output(wait=True)
    plt.close("all")
    display(Markdown("## Investment Opportunity Analysis"))
    display(Markdown(
        "> 💡 **Insight:** Properties with a negative price gap "
        "(actual < model prediction) are *potentially undervalued* — "
        "prime acquisition targets for investors."
    ))

    fig, axes = plt.subplots(1, 2, figsize=(16, 6))

    sns.histplot(df["PriceGap"], kde=True, ax=axes[0], color="mediumseagreen")
    axes[0].axvline(0, color="red", linestyle="--", label="Fair Value")
    axes[0].set_title("Price Gap Distribution\n(Actual − Model Predicted)")
    axes[0].set_xlabel("Price Gap ($)")
    axes[0].legend()

    sns.scatterplot(x="PredictedPrice", y="SalePrice", hue="PriceSegment",
                    data=df, ax=axes[1], alpha=0.5, palette="Set1")
    axes[1].plot([df["SalePrice"].min(), df["SalePrice"].max()],
                 [df["SalePrice"].min(), df["SalePrice"].max()],
                 "r--", label="Perfect Prediction")
    axes[1].set_title("Actual vs Predicted Sale Price")
    axes[1].set_xlabel("Predicted Price ($)")
    axes[1].set_ylabel("Actual Price ($)")
    axes[1].legend()

    plt.tight_layout(); plt.show()

    top_under = (
        df[df["PriceGap"] < 0]
        .nsmallest(10, "PriceGap")
        [["Neighborhood","GrLivArea","OverallQual",
          "SalePrice","PredictedPrice","PriceGap"]]
        .reset_index(drop=True)
    )
    display(Markdown("### 🏆 Top 10 Undervalued Properties"))
    display(top_under)

# ============================================================
# 🎯  TASK 8: MARKET SEGMENTATION — CLUSTERING
# ============================================================
def segmentation():
    clear_output(wait=True)
    plt.close("all")
    display(Markdown("## Task 8 — Market Segmentation (Buyer Personas)"))

    X_sc = StandardScaler().fit_transform(df[CLUSTER_FEATURES])
    inertias, silhouettes = [], []
    K = range(2, 9)
    for k in K:
        km = KMeans(n_clusters=k, random_state=RANDOM_STATE, n_init=10).fit(X_sc)
        inertias.append(km.inertia_)
        silhouettes.append(silhouette_score(X_sc, km.labels_))

    best_k = K[np.argmax(silhouettes)]
    display(Markdown(
        f"> 💡 **Insight:** Silhouette score peaks at **k={best_k}**. "
        f"Clusters represent distinct buyer personas: "
        "Budget · Family · Premium · Luxury."
    ))

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    axes[0].plot(K, inertias, marker="o", color="steelblue")
    axes[0].axvline(N_CLUSTERS, color="red", ls="--", label=f"k={N_CLUSTERS}")
    axes[0].set_title("Elbow Plot"); axes[0].set_xlabel("k"); axes[0].legend()

    axes[1].plot(K, silhouettes, marker="o", color="darkorange")
    axes[1].axvline(N_CLUSTERS, color="red", ls="--", label=f"k={N_CLUSTERS}")
    axes[1].set_title("Silhouette Scores"); axes[1].set_xlabel("k"); axes[1].legend()
    plt.tight_layout(); plt.show()

    fig2, ax2 = plt.subplots(figsize=(13, 6))
    sns.scatterplot(x="GrLivArea", y="SalePrice", hue="Segment",
                    palette="Set2", data=df, ax=ax2, alpha=0.65)
    ax2.set_title(f"Buyer Persona Clusters (k={N_CLUSTERS})")
    ax2.set_xlabel("Living Area (sqft)"); ax2.set_ylabel("Sale Price ($)")
    plt.tight_layout(); plt.show()

    seg_summary = (
        df.groupby("Segment")[["SalePrice","GrLivArea","OverallQual",
                                "PricePerSqFt","TotalBath"]]
        .mean().round(1)
        .rename(columns={
            "SalePrice":    "Avg Price ($)",
            "GrLivArea":    "Avg Area (sqft)",
            "OverallQual":  "Avg Quality",
            "PricePerSqFt": "$/sqft",
            "TotalBath":    "Avg Baths",
        })
    )
    display(Markdown("### Cluster Summary Statistics"))
    display(seg_summary)

# ============================================================
# 🖥️  INTERACTIVE DASHBOARD
# ============================================================
analysis_dict = {
    "📊  Univariate Analysis"          : univariate_analysis,
    "📐  Size & Feature Impact"        : size_analysis,
    "🛏️  Bedrooms & Bathrooms"        : bedroom_bath_analysis,
    "🏠  Amenities & Preferences"      : amenities_analysis,
    "📍  Location Premium"             : location_analysis,
    "📈  Market Trends & Seasonality"  : market_trend,
    "🔗  Correlation Analysis"         : correlation_analysis,
    "🧮  Regression Model"             : regression_model,
    "🔬  VIF — Multicollinearity"      : vif_check,
    "🌲  Feature Importance (RF)"      : feature_importance,
    "💰  Investment Opportunities"     : investment_analysis,
    "🎯  Market Segmentation"          : segmentation,
}

header = widgets.HTML("""
<div style="background:#1a1a2e;color:#e0e0e0;padding:18px 24px;
            border-radius:10px;margin-bottom:12px;font-family:Georgia,serif;">
  <h2 style="margin:0;font-size:22px;letter-spacing:1px;">
    🏡 Real Estate Price Intelligence System
  </h2>
  <p style="margin:4px 0 0 0;font-size:13px;color:#aaa;">
    NextHikes IT Solutions · EDA Project · Sumeet Rajput
  </p>
</div>
""")

dropdown = widgets.Dropdown(
    options=list(analysis_dict.keys()),
    description="Analysis:",
    layout=widgets.Layout(width="70%"),
    style={"description_width": "initial"},
)

output = widgets.Output()

def _on_change(change):
    with output:
        analysis_dict[change["new"]]()

dropdown.observe(_on_change, names="value")
display(header, dropdown, output)
_on_change({"new": list(analysis_dict.keys())[0]})
```
